# Basic Barplot with plotnine

This page ports a basic
[barplot](https://www.data-to-viz.com/graph/barplot.html) example to
[plotnine](https://plotnine.org). The data contain one numeric value for each
category.

The source example is the
[Basic Barplot](https://python-graph-gallery.com/1-basic-barplot/) tutorial from
the [Python Graph Gallery](https://python-graph-gallery.com), which uses
matplotlib directly.

## Libraries and data

The data are stored in **pandas**. The plot uses the core plotnine bar geometry.

```python
import pandas as pd
from plotnine import (
    ggplot,
    aes,
    geom_col,
    labs,
    theme_minimal,
)

# Create the dataset (same data as the original matplotlib example)
df = pd.DataFrame({
    "category": ["A", "B", "C", "D", "E"],
    "height": [3, 12, 5, 18, 45],
})
```

## Basic barplot

A bar plot with precomputed heights uses two pieces:

1. **`ggplot(df, aes(...))`** binds the data and maps columns to aesthetics.
2. **`geom_col()`** draws bars whose heights come from the data. Use
   `geom_bar()` when plotnine should count rows instead.

```python
(
    ggplot(df, aes(x="category", y="height"))
    + geom_col()
)
```

## Labels and theme

The same plot can add a fixed fill colour, axis labels, a title, and a theme.

```python
(
    ggplot(df, aes(x="category", y="height"))
    + geom_col(fill="#69b3a2")
    + labs(
        title="Basic Barplot with plotnine",
        x="Category",
        y="Value",
    )
    + theme_minimal()
)
```

## Related features

Common extensions are:

* Map `fill` to a column for grouped or stacked bars.
* Use `coord_flip()` for horizontal bars.
* Add uncertainty with `geom_errorbar()`.
* Split panels with `facet_wrap()` or `facet_grid()`.
